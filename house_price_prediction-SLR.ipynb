```python
df = df.withColumnRenamed("number of bedrooms", "bhk") \
       .withColumnRenamed("number of bathrooms", "bath") \
       .withColumnRenamed("living area", "total_sqft") \
       .withColumnRenamed("Price", "price")
```

```python
required_cols = ["bhk", "bath", "total_sqft", "Postal Code", "price"]
df_clean = df.dropna(subset=required_cols)
```

```python
from pyspark.ml.feature import StringIndexer

indexer = StringIndexer(inputCol="Postal Code", outputCol="location_indexed")
df_indexed = indexer.fit(df_clean).transform(df_clean)
```

```python
df_indexed.select("bhk", "bath", "total_sqft","location_indexed", "price").show(10)
```

```python
from pyspark.ml.feature import VectorAssembler

feature_cols = ["bhk", "bath", "total_sqft", "location_indexed"]
assembler = VectorAssembler(inputCols=feature_cols, outputCol="features")

final_df = assembler.transform(df_indexed)
final_df.select("features", "price").show(5, truncate=False)
```

```python
from pyspark.ml.regression import LinearRegression

# Split into training and test data
train_data, test_data = final_df.randomSplit([0.8, 0.2], seed=42)

# Initialize model
lr = LinearRegression(featuresCol="features", labelCol="price")

# Fit model
lr_model = lr.fit(train_data)

# Evaluate on test data
test_results = lr_model.evaluate(test_data)

# Show evaluation metrics
print("RMSE:", test_results.rootMeanSquaredError)
print("R2:", test_results.r2)
```

```python
predictions = lr_model.transform(test_data)
predictions.select("features", "price", "prediction").show(10, truncate=False)
```

```python
predictions = lr_model.transform(test_data)
predictions = predictions.withColumn("residual", predictions["price"] - predictions["prediction"])
predictions.select("price", "prediction", "residual").show(10)
```
